Run this cell first to confirm the interpreter, platform, and core package

````python
from pathlib import Path
import subprocess
import sys

NOTEBOOK_SUBDIR = "notebooks"
COLAB_PACKAGES = {}
REPO_NAME = "py4fi3rd"
REPO_URL = "https://github.com/yhilpisch/py4fi3rd.git"


def _support_dir() -> Path:
    cwd = Path.cwd().resolve()
    for candidate in (cwd, *cwd.parents):
        support_dir = candidate / "notebooks"
        if (support_dir / "_book_notebook_support.py").exists():
            return support_dir
    if "google.colab" in sys.modules:
        root = Path("/content") / REPO_NAME
        if not root.exists():
            subprocess.run(
                ["git", "clone", "--depth", "1", REPO_URL, str(root)],
                check=True,
            )
        return root / "notebooks"
    raise RuntimeError("Could not locate notebook support helpers.")


SUPPORT_DIR = _support_dir()
if str(SUPPORT_DIR) not in sys.path:
    sys.path.insert(0, str(SUPPORT_DIR))

from _book_notebook_support import setup_notebook

CONTEXT = setup_notebook(
    notebook_subdir=NOTEBOOK_SUBDIR,
    colab_packages=COLAB_PACKAGES,
)

PROJECT_ROOT = CONTEXT["PROJECT_ROOT"]
NOTEBOOK_DIR = CONTEXT["NOTEBOOK_DIR"]
CODE_DIR = CONTEXT["CODE_DIR"]
CHAPTERS_DIR = CONTEXT["CHAPTERS_DIR"]
FIGURES_DIR = CONTEXT["FIGURES_DIR"]
DATA_DIR = CONTEXT["DATA_DIR"]

PROJECT_ROOT
````

# Installing Python


The first check is whether your interpreter matches the version and
installation path you expect.


## Linux and macOS


Unix-like systems usually rely on package managers or official installers, so
verify which path you are using.


```bash
python3 --version
```


## Windows


On Windows, verify your launcher, terminal, and environment activation
commands.


# Working with the Shell


A shell session is where you confirm paths, run scripts, and activate
environments.


## Shells on Linux and macOS


Use `bash`, `zsh`, or a comparable shell to manage Python projects on Unix-
like systems.


```bash
user@machine:~$
```


## Shells on Windows


PowerShell and Windows Terminal provide the same command-line workflow on
Windows.


# Managing Environments with venv and pip


Use `venv` and `pip` to isolate dependencies for each project.


## Creating a Virtual Environment


A virtual environment gives each notebook or script its own dependency set.


```bash
python3 -m venv .venv
```


```bash
source .venv/bin/activate
```


## Installing Packages with pip


Install only the packages your project needs, and keep them inside the active
environment.


```bash
python -m pip install numpy pandas matplotlib
```


```bash
deactivate
```


# Editors on the Shell


A terminal editor is enough for quick edits, configuration files, and small
scripts.


## Getting Started with nano


Use `nano` when you want a minimal editor with visible keybindings.


```bash
nano script.py
```


## Using vim as a Portable Power Tool


`vim` pays off when you work on remote machines or need fast text editing.


```bash
vim script.py
```
