# Environment and Version Check
Run this cell first to confirm the interpreter, platform, and core package
versions that the notebook will use.


````python
import os
import platform
import subprocess
import sys
from importlib import metadata
from pathlib import Path


PACKAGES = [
    "pip",
    "setuptools",
    "wheel",
    "ipykernel",
    "jupyterlab",
    "numpy",
    "pandas",
    "matplotlib",
]


def print_package_version(name: str) -> None:
    """Print one installed package version."""

    try:
        version = metadata.version(name)
    except metadata.PackageNotFoundError:
        version = "not installed"
    print(f"{name}: {version}")


print("Python executable:", sys.executable)
print("Python version:", sys.version.replace("\n", " "))
print("Python version info:", sys.version_info[:])
print("Platform:", platform.platform())
print("Machine:", platform.machine())
print("Processor:", platform.processor())
print("Hostname:", platform.node())
print("Current working directory:", Path.cwd())
print("Virtual env:", os.environ.get("VIRTUAL_ENV", "<none>"))
print("Conda env:", os.environ.get("CONDA_DEFAULT_ENV", "<none>"))
print("PATH head:")
for entry in os.environ.get("PATH", "").split(os.pathsep)[:5]:
    print("  ", entry)

print("Installed package versions:")
for package in PACKAGES:
    print_package_version(package)

pip_cmd = [sys.executable, "-m", "pip", "--version"]
pip_info = subprocess.run(
    pip_cmd, capture_output=True, text=True, check=False
)
print("pip command:", pip_info.stdout.strip())
````

# Managing Environments with venv and pip


Use `venv` and `pip` to isolate dependencies for each project.


## Creating a Virtual Environment


A virtual environment gives each notebook or script its own dependency set.


```bash
python3 -m venv .venv
```


```bash

```


## Installing Packages with pip


Install only the packages your project needs, and keep them inside the active
environment.


```bash
python -m pip install numpy pandas matplotlib
```


```bash
deactivate
```
