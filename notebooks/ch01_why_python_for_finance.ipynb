## Minimal Trade Representation
The chapter script uses a tiny trade object to show how Python
turns financial ideas into reproducible code.


```python
from dataclasses import dataclass
from typing import Iterable


@dataclass
class SimpleTrade:
    """Minimal representation of a trade for reporting."""

    symbol: str
    qty: float
    price: float

    def notional(self) -> float:
        """Return signed notional value."""

        return self.qty * self.price


def portfolio_notional(trades: Iterable[SimpleTrade]) -> float:
    """Compute aggregate signed notional."""

    return float(sum(trade.notional() for trade in trades))


trades = [
    SimpleTrade("AAPL", 10, 180.0),
    SimpleTrade("MSFT", -5, 350.0),
]
portfolio_notional(trades)  # Combine long and short notionals.
```
