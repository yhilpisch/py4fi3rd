```python
import pandas as pd


def summarize_prices(data: pd.DataFrame) -> dict[str, float]:
    """Summarize close prices with validation."""

    if "close" not in data.columns:  # Require the input column.
        raise ValueError(
            "expected a 'close' column in the input DataFrame"
        )
    close = pd.to_numeric(
        data["close"], errors="coerce"
    ).dropna()  # Keep valid numeric prices.
    return {
        "mean": float(close.mean()),
        "median": float(close.median()),
        "std": float(close.std(ddof=1)),
    }
```

```python
sample = pd.DataFrame({"close": [100.4, 101.1, 102.19]})
summarize_prices(sample)
```
